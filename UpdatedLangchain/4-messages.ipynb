### Messages

Messages are the fundamental unit of context for models in LangChain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation with intercating with an LLM. Messages are objects that contain:

1. Role - Identifies the user type (e.g. system, user)
2. Content - Represents the actual content of the message (like text, images and audio, documents etc.)
3. Metadata - Optional fields such as response information, message IDs, and token usage

LangChain provides a standared message type that works across all model providers, ensuring consistent behaviour regardless of the model being called.

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model=init_chat_model(model="gpt-5")
model
```

```python
model.invoke("Please tell what is AI")
```

### Text Prompts

Text prompts are strings - ideal for straightforward generation tasks where you don't need to retain conversation history.

```python
model.invoke("What is LangChain")
```

Use text prompts when:
1. You have a single, standalone request
2. You don't need conversation history
3. You want minimal code complexity

### Message Prompts

Alternatively, you can pass in a list of messages to the model by providing a list of message objects.

Message Types:
1. System Message - Tells the model how to behave and provide context for interactions
2. Human Message - Represents user imput and interactions with the model
3. AI Message - Responses generated by the model, including text context, tool calls, and metadata
4. Tool Message - Represents the outputs of tool calls

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage(content="You are a poetry expert."),
    HumanMessage(content="Write a poem about AI."),
    ]

response = model.invoke(messages)
response.content
```

```python
system_msg = SystemMessage(content="You are a helpful coding assistant.")

messages = [
    system_msg,
    HumanMessage("How do i create a rest api.")]

response = model.invoke(messages)
response.content
```

```python
## Message metadata

human_msg = HumanMessage(content="Hello", name="alice", id="msg_123")
```

```python
response = model.invoke([human_msg])
response
```
